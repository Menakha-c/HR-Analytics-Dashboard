```python
#Feature Engineering
```

```python
#Step 1: Create Age Group
```

```python
df["Age"].describe()
```

```python
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 30, 45, 60],
    labels=["Young", "Middle-aged", "Senior"]
)
```

```python
df[["Age", "AgeGroup"]].head(10)
```

```python
#Step 2: Create Experience Group
```

```python
df["ExperienceGroup"] = pd.cut(
    df["TotalWorkingYears"],
    bins=[-1, 5, 10, 20, 40],
    labels=["0-5 Years", "6-10 Years", "11-20 Years", "20+ Years"]
)
```

```python
df[["TotalWorkingYears", "ExperienceGroup"]].head(10)
```

```python
#Step 3: Create Income Category
```

```python
df["IncomeCategory"] = pd.qcut(
    df["MonthlyIncome"],
    q=3,
    labels=["Low", "Medium", "High"]
)
```

```python
df[["MonthlyIncome", "IncomeCategory"]].head(10)
```

```python
#Step 4: Create Attrition Flag
```

```python
df["AttritionFlag"] = df["Attrition"].map({
    "Yes": 1,
    "No": 0
})
```

```python
df[["Attrition", "AttritionFlag"]].head()
```
